## 4. Метрики качества

Метрики рассчитаны по 19 объектам, для которых модель выдала числовой прогноз. 20-й ответ (ошибка преобразования строки в число) не является числовой парой «факт–прогноз», поэтому в MAE, RMSE и R² он не включается.

**MAE = 580 079,631579 TRY**

**RMSE = 774 590,486239 TRY**

**R² = −0,492099**

Ручной расчёт и расчёт стандартными методами дали одинаковые значения с учётом округления.

## 5. Анализ результатов

Средняя абсолютная ошибка составляет около 580 тыс. TRY, а RMSE — около 775 тыс. TRY. Более высокое значение RMSE показывает наличие крупных ошибок прогнозирования.

R² имеет отрицательное значение (−0,492099), поэтому на этих 19 числовых результатах прогнозирование хуже базового сравнения с постоянным прогнозом, равным среднему фактическому значению.

Особенно заметны большие расхождения на объектах с низкой фактической ценой: например, при фактических значениях 3 500–16 000 TRY модель выдавала прогнозы значительно выше. Такие ошибки существенно увеличивают MAE и RMSE.

На объекте id 1 модель вернула не число, а ошибку «could not convert string to float». Поскольку ошибка является фактическим ответом модели, она сохранена в таблице и CSV. Для полного числового тестирования этот объект требует исправления этапа преобразования входных данных.

## 6. Вывод

На предоставленных тестовых результатах модель сформировала 19 числовых прогнозов и 1 ошибку. Числовые метрики указывают на значительное расхождение прогнозов с фактическими значениями. Для дальнейшей проверки необходимо исправить наши модели прогнозирования данных (увеличить точность) и снова повторить тестирование всех 20 объектов.
